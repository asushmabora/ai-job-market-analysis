```python
df['required_skills'].head(10)
```

```python
skills_list = df['required_skills'].str.split('|').explode().tolist()
print("Step 1:", skills_list)
```

```python
unique_skills = pd.Series(skills_list).drop_duplicates().tolist()
```

```python
##skill table
skills_table = pd.DataFrame({
    'skill_id': range(1, len(unique_skills)+1),
    'skill_name': unique_skills
})
print("Step 3:\n", skills_table)
```

```python
lookup = dict(zip(skills_table['skill_name'], skills_table['skill_id']))
print("Step 4:", lookup)
```

```python
bridge_rows = []
for _, row in df.iterrows():
    job = row['job_id']
    for skill in row['required_skills'].split('|'):
        bridge_rows.append({'job_id': job, 'skill_id': lookup[skill]})

bridge_table = pd.DataFrame(bridge_rows)
print("Step 5:\n", bridge_table)
```

```python
print("locations_df.head():\n", locations_df.head())
print("locations_df.shape:", locations_df.shape)

print("\nskills_df.head():\n", skills_table.head())
print("skills_df.shape:", skills_table.shape)

```

```python
jobs_df = df.copy()
```

```python
jobs_df = jobs_df.merge(
    locations_df[['location_id','city','country']],
    on=['city','country'],
    how='left'
)
```

```python
jobs_df = jobs_df.drop(columns=[
    'city',
    'country',
    'required_skills',
    'demand_score',
    'demand_growth_yoy_pct',
    'benefits_score_10',
    'ai_salary_premium_pct'
])
```

```python
# job skills table
job_skills = df[['job_id', 'required_skills']].copy()
```

```python
job_skills['required_skills'] = job_skills['required_skills'].str.split('|')
job_skills = job_skills.explode('required_skills')
```

```python
job_skills_df = job_skills.merge(
    skills_table,
    left_on='required_skills',
    right_on='skill_name',
    how='left'
)
```

```python
#  Keep only job_id and skill_id
job_skills_df = job_skills_df[['job_id', 'skill_id']]
```

```python
job_skills_df = job_skills_df.reset_index(drop=True)
print("job_skills_df.head():\n", job_skills_df.head())
```

```python
job_skills_df = job_skills_df.drop_duplicates().reset_index(drop=True)
print("Duplicates after cleanup:", job_skills_df.duplicated().sum())
```
